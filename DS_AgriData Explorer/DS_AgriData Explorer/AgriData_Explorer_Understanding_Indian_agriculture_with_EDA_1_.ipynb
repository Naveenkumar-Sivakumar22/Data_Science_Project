```python
# Identify numerical columns for outlier detection
numerical_cols_for_outliers = df_last_50_years.select_dtypes(include=['float64']).columns.tolist()

initial_rows_outliers = df_last_50_years.shape[0]
print(f"Initial number of rows before outlier removal: {initial_rows_outliers}")

df_no_outliers = df_last_50_years.copy()

# Apply IQR-based outlier removal for each numerical column
for col in numerical_cols_for_outliers:
    Q1 = df_no_outliers[col].quantile(0.25)
    Q3 = df_no_outliers[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Filter out outliers
    df_no_outliers = df_no_outliers[(df_no_outliers[col] >= lower_bound) & (df_no_outliers[col] <= upper_bound)]

final_rows_outliers = df_no_outliers.shape[0]
print(f"Number of rows after outlier removal: {final_rows_outliers}")
print(f"Number of rows removed: {initial_rows_outliers - final_rows_outliers}")

print("\nFirst 5 rows of the DataFrame after outlier removal:")
display(df_no_outliers.head())

print("\nDataFrame Info after outlier removal:")
df_no_outliers.info()
```

```python
import pandas as pd
import sqlite3

# Load the data from cleaned_data_1.csv specifically for SQL querying
# This ensures 'Year' and 'State Name' are available and other transformations are consistent.
df_for_sql_query = pd.read_csv('/content/cleaned_data_1.csv')

# Re-apply the column unit conversions for consistency, as done in previous steps.
# Convert 'AREA (1000 ha)' columns to 'AREA (ha)' and update values
for col in df_for_sql_query.columns:
    if 'AREA (1000 ha)' in col:
        df_for_sql_query[col] = df_for_sql_query[col] * 1000
        new_col_name = col.replace('AREA (1000 ha)', 'AREA (ha)')
        df_for_sql_query.rename(columns={col: new_col_name}, inplace=True)

# Convert 'PRODUCTION (1000 tons)' columns to 'PRODUCTION (kg)' and update values
for col in df_for_sql_query.columns:
    if 'PRODUCTION (1000 tons)' in col:
        df_for_sql_query[col] = df_for_sql_query[col] * 1000000
        new_col_name = col.replace('PRODUCTION (1000 tons)', 'PRODUCTION (kg)')
        df_for_sql_query.rename(columns={col: new_col_name}, inplace=True)

# Create an in-memory SQLite database connection
conn = sqlite3.connect(':memory:')

# Write the DataFrame to a SQL table named 'crop_data'
df_for_sql_query.to_sql('crop_data', conn, index=False, if_exists='replace')

# Step 1: Identify the top 3 states by total RICE PRODUCTION (kg)
top_3_states_query = """
SELECT
    "State Name",
    SUM("RICE PRODUCTION (kg)") AS Total_Rice_Production_kg
FROM
    crop_data
GROUP BY
    "State Name"
ORDER BY
    Total_Rice_Production_kg DESC
LIMIT 3
"""
top_3_states_df = pd.read_sql_query(top_3_states_query, conn)
top_3_state_names = top_3_states_df['State Name'].tolist()

# Convert the list of state names to a comma-separated string for the SQL IN clause
top_3_state_names_str = ", ".join(f"'{state}'" for state in top_3_state_names)

# Step 2: Get year-wise trend for these top 3 states
sql_query_trend = f"""
SELECT
    Year,
    "State Name",
    SUM("RICE PRODUCTION (kg)") AS Year_Wise_Rice_Production_kg
FROM
    crop_data
WHERE
    "State Name" IN ({top_3_state_names_str})
GROUP BY
    Year,
    "State Name"
ORDER BY
    Year ASC,
    "State Name" ASC
"""

rice_production_trend = pd.read_sql_query(sql_query_trend, conn)

print("Year-wise Trend of Rice Production Across Top 3 States:")
display(rice_production_trend.head(10)) # Display first 10 rows for brevity

# Close the database connection
conn.close()

# Save the result to a CSV file
output_file_name_sql = 'rice_production_trend_output_1.csv'
rice_production_trend.to_csv(output_file_name_sql, index=False)
print(f"SQL query result saved to {output_file_name_sql} successfully.")
print("You can now download this file from the Colab file browser.")
```

```python
import pandas as pd
import sqlite3

# Load the data from cleaned_data_1.csv specifically for SQL querying
df_for_sql_query = pd.read_csv('/content/cleaned_data_1.csv')

# Re-apply the column unit conversions for consistency, as done in previous steps.
# Convert 'AREA (1000 ha)' columns to 'AREA (ha)' and update values
for col in df_for_sql_query.columns:
    if 'AREA (1000 ha)' in col:
        df_for_sql_query[col] = df_for_sql_query[col] * 1000
        new_col_name = col.replace('AREA (1000 ha)', 'AREA (ha)')
        df_for_sql_query.rename(columns={col: new_col_name}, inplace=True)

# Convert 'PRODUCTION (1000 tons)' columns to 'PRODUCTION (kg)' and update values
for col in df_for_sql_query.columns:
    if 'PRODUCTION (1000 tons)' in col:
        df_for_sql_query[col] = df_for_sql_query[col] * 1000000
        new_col_name = col.replace('PRODUCTION (1000 tons)', 'PRODUCTION (kg)')
        df_for_sql_query.rename(columns={col: new_col_name}, inplace=True)

# Create an in-memory SQLite database connection
conn = sqlite3.connect(':memory:')

# Write the DataFrame to a SQL table named 'crop_data'
df_for_sql_query.to_sql('crop_data', conn, index=False, if_exists='replace')

major_crops = ['RICE', 'WHEAT', 'MAIZE']
all_correlations = []

for crop in major_crops:
    area_col = f'{crop} AREA (ha)'
    production_col = f'{crop} PRODUCTION (kg)'

    # Construct SQL query to fetch relevant data
    sql_query_crop_data = f"""
    SELECT
        "Dist Name",
        "{area_col}" AS Area,
        "{production_col}" AS Production
    FROM
        crop_data
    WHERE
        Area IS NOT NULL AND Production IS NOT NULL
        AND Area > 0 AND Production > 0
    """

    # Fetch data into a pandas DataFrame
    crop_data_df = pd.read_sql_query(sql_query_crop_data, conn)

    # Calculate district-wise correlation in pandas
    for dist_name, group in crop_data_df.groupby('Dist Name'):
        # Correlation requires at least 2 data points for both variables
        if len(group) > 1 and group['Area'].std() > 0 and group['Production'].std() > 0:
            correlation = group['Area'].corr(group['Production'])
            all_correlations.append({
                'Dist Name': dist_name,
                'Crop': crop,
                'Correlation': correlation
            })

# Close the database connection
conn.close()

# Convert results to a DataFrame
correlation_df = pd.DataFrame(all_correlations)

print("District-wise Correlation between Area and Production for Major Crops:")
display(correlation_df.head())

# Save the result to a CSV file
output_file_name_correlation = 'District-wise Correlation Between Area and Production for Major Crops (Rice, Wheat, and Maize).csv'
correlation_df.to_csv(output_file_name_correlation, index=False)
print(f"\nCorrelation data saved to {output_file_name_correlation} successfully.")
print("You can now download this file from the Colab file browser.")
```

```python
import pandas as pd
import sqlite3

# Load the data from cleaned_data_1.csv specifically for SQL querying
df_for_sql_query = pd.read_csv('/content/cleaned_data_1.csv')

# Re-apply the column unit conversions for consistency
for col in df_for_sql_query.columns:
    if 'AREA (1000 ha)' in col:
        df_for_sql_query[col] = df_for_sql_query[col] * 1000
        new_col_name = col.replace('AREA (1000 ha)', 'AREA (ha)')
        df_for_sql_query.rename(columns={col: new_col_name}, inplace=True)

for col in df_for_sql_query.columns:
    if 'PRODUCTION (1000 tons)' in col:
        df_for_sql_query[col] = df_for_sql_query[col] * 1000000
        new_col_name = col.replace('PRODUCTION (1000 tons)', 'PRODUCTION (kg)')
        df_for_sql_query.rename(columns={col: new_col_name}, inplace=True)

# Create an in-memory SQLite database connection
conn = sqlite3.connect(':memory:')

# Write the DataFrame to a SQL table named 'crop_data'
df_for_sql_query.to_sql('crop_data', conn, index=False, if_exists='replace')

# Step 1: Identify the top 5 states by total Cotton Production
top_5_states_cotton_query = """
SELECT
    "State Name",
    SUM("COTTON PRODUCTION (kg)") AS Total_Cotton_Production_kg
FROM
    crop_data
GROUP BY
    "State Name"
ORDER BY
    Total_Cotton_Production_kg DESC
LIMIT 5
"""
top_5_states_df = pd.read_sql_query(top_5_states_cotton_query, conn)
top_5_state_names = top_5_states_df['State Name'].tolist()

# Convert the list of state names to a comma-separated string for the SQL IN clause
top_5_state_names_str = ", ".join(f"'{state}'" for state in top_5_state_names)

# Step 2: Calculate yearly production growth for these top 5 states
sql_query_cotton_growth = f"""
WITH YearlyProduction AS (
    SELECT
        Year,
        "State Name",
        SUM("COTTON PRODUCTION (kg)") AS CurrentYearProduction
    FROM
        crop_data
    WHERE
        "State Name" IN ({top_5_state_names_str})
    GROUP BY
        Year,
        "State Name"
),
LaggedProduction AS (
    SELECT
        Year,
        "State Name",
        CurrentYearProduction,
        LAG(CurrentYearProduction, 1, 0) OVER (PARTITION BY "State Name" ORDER BY Year) AS PreviousYearProduction
    FROM
        YearlyProduction
)
SELECT
    Year,
    "State Name",
    CurrentYearProduction,
    PreviousYearProduction,
    CASE
        WHEN PreviousYearProduction > 0 THEN ((CurrentYearProduction - PreviousYearProduction) * 100.0 / PreviousYearProduction)
        ELSE 0 -- Handle cases where previous year production was zero or no data
    END AS GrowthRatePercentage
FROM
    LaggedProduction
ORDER BY
    "State Name", Year
"""

cotton_production_growth = pd.read_sql_query(sql_query_cotton_growth, conn)

print("Yearly Cotton Production Growth for Top 5 States:")
display(cotton_production_growth.head(10)) # Display first 10 rows for brevity

# Close the database connection
conn.close()

# Save the result to a CSV file
output_file_name_growth = 'Yearly Production Growth of Cotton in Top 5 Cotton Producing States.csv'
cotton_production_growth.to_csv(output_file_name_growth, index=False)
print(f"\nSQL query result saved to {output_file_name_growth} successfully.")
print("You can now download this file from the Colab file browser.")
```
