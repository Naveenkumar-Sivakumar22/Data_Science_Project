```python
def parse_rating_count(count_str):
    if isinstance(count_str, (int, float)):
        return float(count_str)
    
    if count_str == 'Too Few Ratings':
        return 0
    elif isinstance(count_str, str) and '+' in count_str:
        count_str = count_str.replace(' ratings', '').replace('+', '')
        if 'K' in count_str:
            return float(count_str.replace('K', '')) * 1000
        else:
            return float(count_str)
    
    try:
        return float(count_str)
    except (ValueError, TypeError):
        return 0  

df['rating_count'] = df['rating_count'].apply(parse_rating_count)

print("Data type of 'rating_count' after conversion:", df['rating_count'].dtype)
print("Value counts for 'rating_count' after conversion (top 10):\n", df['rating_count'].value_counts().head(10))
```

```python
from sklearn.preprocessing import MultiLabelBinarizer
import pandas as pd
from typing import List

def encode_cuisine_column(df: pd.DataFrame, cuisine_col: str = 'cuisine') -> pd.DataFrame:
    """
    One-hot encodes the cuisine column which contains comma-separated values.
    
    Args:
        df (pd.DataFrame): Input DataFrame containing the cuisine column
        cuisine_col (str, optional): Name of the column containing cuisine data. Defaults to 'cuisine'.
    
    Returns:
        pd.DataFrame: Original DataFrame with added one-hot encoded cuisine columns
    """
    # Clean and split the cuisine strings
    cuisine_lists = df[cuisine_col].str.split(',').apply(lambda x: [item.strip() for item in x])
    
    # Initialize and fit the MultiLabelBinarizer
    mlb = MultiLabelBinarizer()
    cuisine_encoded = pd.DataFrame(
        mlb.fit_transform(cuisine_lists),
        columns=[f"cuisine_{cls}" for cls in mlb.classes_],  # Add 'cuisine_' prefix to avoid column name conflicts
        index=df.index
    )
    
    # Join the encoded columns to the original DataFrame
    result_df = pd.concat([df, cuisine_encoded], axis=1)  # Using concat instead of join for more control
    
    # Display information about the transformation
    print(f"Shape of DataFrame after cuisine encoding: {result_df.shape}")
    print("First 5 rows of encoded cuisines:")
    print(cuisine_encoded.head())  # Show only the encoded part
    
    return result_df

# Apply the function to encode the cuisine column
df = encode_cuisine_column(df)
```

```python
import pickle
from typing import Any
import os
from sklearn.preprocessing import MultiLabelBinarizer  # Import the MultiLabelBinarizer class

def save_pickle_object(obj: Any, filepath: str) -> None:
    """
    Save an object to a pickle file.
    
    Args:
        obj (Any): The object to save.
        filepath (str): Path where the pickle file will be saved.
        
    Raises:
        IOError: If there's an issue writing to the file.
    """
    try:
        # Create directory if it doesn't exist
        directory = os.path.dirname(filepath)
        if directory and not os.path.exists(directory):
            os.makedirs(directory)
            
        # Save the object
        with open(filepath, 'wb') as file:
            pickle.dump(obj, file)
        print(f"Object successfully saved to '{filepath}'")
    except Exception as e:
        print(f"Error saving object to '{filepath}': {str(e)}")
        raise

# Create and fit a MultiLabelBinarizer instance before saving it
mlb = MultiLabelBinarizer()  # Define the mlb variable
# Typically you would fit the binarizer with some data here
# mlb.fit(your_data)

# Save the MultiLabelBinarizer
save_pickle_object(mlb, 'encoder.pkl')
```

```python
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.preprocessing import MinMaxScaler
from sklearn.neighbors import NearestNeighbors

# 1. Load Data
df = pd.read_csv('cleaned_data.csv', on_bad_lines='skip')

# 2. Preprocessing & Encoding
# Clean 'cost': Remove '₹' and whitespace, convert to numeric
df['cost_clean'] = df['cost'].astype(str).str.replace('₹', '').str.strip()
df['cost_clean'] = pd.to_numeric(df['cost_clean'], errors='coerce').fillna(0)

# Clean 'rating': Convert to numeric, handle '--' or errors
df['rating_clean'] = pd.to_numeric(df['rating'], errors='coerce').fillna(0)

# Encode 'cuisine': One-Hot Encoding (handling comma-separated values)
cuisine_dummies = df['cuisine'].str.get_dummies(sep=',')

# Normalize numerical features
scaler = MinMaxScaler()
numerical_features = scaler.fit_transform(df[['cost_clean', 'rating_clean']])
numerical_df = pd.DataFrame(numerical_features, columns=['cost_norm', 'rating_norm'], index=df.index)

# Combine all features into the final encoded dataframe
df_encoded = pd.concat([numerical_df, cuisine_dummies], axis=1)

print(f"Data Processed. Features shape: {df_encoded.shape}")

# 3. Use NearestNeighbors instead of computing the full similarity matrix
# This is much more memory efficient for large datasets
model = NearestNeighbors(n_neighbors=6, algorithm='brute', metric='cosine')  # 6 because we include the restaurant itself
model.fit(df_encoded)

# 4. Define Recommendation Function
def get_recommendations(restaurant_name, df, model, top_n=5):
    """
    Finds similar restaurants using NearestNeighbors model
    """
    # Check if restaurant exists
    if restaurant_name not in df['name'].values:
        return f"Restaurant '{restaurant_name}' not found."

    # Get the index of the restaurant
    idx = df[df['name'] == restaurant_name].index[0]
    
    # Get the features for this restaurant
    restaurant_features = df_encoded.iloc[idx].values.reshape(1, -1)
    
    # Find nearest neighbors
    distances, indices = model.kneighbors(restaurant_features)
    
    # Convert distances to similarity scores (1 - distance)
    similarity_scores = 1 - distances.flatten()
    
    # Skip the first result (the restaurant itself)
    top_indices = indices.flatten()[1:top_n+1]
    top_scores = similarity_scores[1:top_n+1]

    # Map back to original data
    result = df.iloc[top_indices][['name', 'city', 'cuisine', 'cost', 'rating']].copy()
    result['similarity_score'] = top_scores  # Add score for reference

    return result

# 5. Usage Example
sample_restaurant = df['name'].iloc[0]
print(f"\n--- Recommendations for: {sample_restaurant} ---")

recommendations = get_recommendations(sample_restaurant, df, model)
print(recommendations)
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- 1. Page Configuration ---
st.set_page_config(page_title="Restaurant Recommender", layout="wide", page_icon="🍴")

# --- 2. Data Loading & Cleaning ---
@st.cache_data
def load_data():
    try:
        # Load data
        df = pd.read_csv('cleaned_data.csv')
        
        # Standardize column names (lowercase and no spaces)
        df.columns = [c.strip().lower().replace(" ", "_") for c in df.columns]

        # Clean 'cost' column: Remove '₹' and commas, convert to numeric
        if 'cost' in df.columns:
            df['price'] = df['cost'].str.replace('₹', '', regex=False).str.replace(',', '', regex=False)
            df['price'] = pd.to_numeric(df['price'], errors='coerce')
        
        # Clean 'rating' column: Convert '--' or 'Too Few Ratings' to NaN, then numeric
        if 'rating' in df.columns:
            df['rating_numeric'] = pd.to_numeric(df['rating'], errors='coerce').fillna(0.0)
        
        # Clean 'cuisine' column
        df['cuisine'] = df['cuisine'].fillna('Not Specified')
        
        return df
    except Exception as e:
        st.error(f"Error loading CSV: {e}")
        return pd.DataFrame()

# Initialize Data
df = load_data()

if df.empty:
    st.error("Could not load 'cleaned_data.csv'. Please ensure the file is in the same directory.")
else:
    # --- 3. Sidebar Filters ---
    st.sidebar.header("🔍 Filter Preferences")

    # City Filter
    all_cities = sorted(df['city'].dropna().unique())
    selected_city = st.sidebar.selectbox("Choose a City", options=["All"] + all_cities)

    # Cuisine Filter
    all_cuisines = set()
    df['cuisine'].str.split(',').apply(lambda x: [all_cuisines.add(c.strip()) for c in x if isinstance(x, list)])
    sorted_cuisines = sorted(list(all_cuisines))
    selected_cuisine = st.sidebar.selectbox("Choose a Cuisine", options=["All"] + sorted_cuisines)

    # Price Filter (Strictly capped at 2000 as requested)
    max_price = st.sidebar.slider("Maximum Budget (₹)", min_value=0, max_value=2000, value=500, step=50)

    # Rating Filter
    min_rating = st.sidebar.slider("Minimum Rating", min_value=0.0, max_value=5.0, value=3.5, step=0.1)

    # --- 4. Recommendation Engine Logic ---
    filtered_df = df.copy()

    if selected_city != "All":
        filtered_df = filtered_df[filtered_df['city'] == selected_city]
    
    if selected_cuisine != "All":
        filtered_df = filtered_df[filtered_df['cuisine'].str.contains(selected_cuisine, case=False, na=False)]
        
    filtered_df = filtered_df[filtered_df['price'] <= max_price]
    filtered_df = filtered_df[filtered_df['rating_numeric'] >= min_rating]

    # Sort results by Rating (Best first) then Price (Cheaper first)
    recommendations = filtered_df.sort_values(by=['rating_numeric', 'price'], ascending=[False, True])

    # --- 5. Main Dashboard UI ---
    st.title("🍽️ Restaurant Recommendation Engine")
    st.markdown(f"Find the best dining spots in **{selected_city}** under **₹{max_price}**.")

    # Top Metrics
    m1, m2, m3, m4 = st.columns(4)
    m1.metric("Total Matches", len(recommendations))
    m2.metric("Avg Rating", f"{recommendations['rating_numeric'].mean():.1f} ⭐" if not recommendations.empty else "N/A")
    m3.metric("Market Avg Cost", f"₹{df['price'].mean():.0f}")
    m4.metric("Selection Avg Cost", f"₹{recommendations['price'].mean():.0f}" if not recommendations.empty else "N/A")

    st.divider()

    # Tabs for different views
    tab1, tab2, tab3 = st.tabs(["✨ Top Picks", "📋 Comparison Table", "📈 Market Analysis"])

    with tab1:
        if not recommendations.empty:
            # Highlight the #1 Recommendation
            top_pick = recommendations.iloc[0]
            st.subheader("Our Top Recommendation")
            with st.container(border=True):
                c1, c2 = st.columns([3, 1])
                with c1:
                    st.markdown(f"### {top_pick['name']}")
                    st.write(f"📍 **Address:** {top_pick.get('address', 'N/A')}")
                    st.write(f"🍴 **Cuisine:** {top_pick['cuisine']}")
                with c2:
                    st.metric("Rating", f"⭐ {top_pick['rating']}")
                    st.metric("Cost for Two", f"₹{top_pick['price']}")

            st.write("---")
            st.subheader("Other Great Options")
            # Show next top 5
            for i in range(1, min(6, len(recommendations))):
                row = recommendations.iloc[i]
                with st.expander(f"{row['name']} — ⭐ {row['rating']} — ₹{row['price']}"):
                    st.write(f"**Cuisine:** {row['cuisine']}")
                    st.write(f"**Location:** {row['city']}")
                    st.write(f"**Address:** {row.get('address', 'N/A')}")
        else:
            st.error("No restaurants found matching your criteria. Try reducing the rating or increasing the budget.")

    with tab2:
        st.subheader("Restaurant Comparison")
        if not recommendations.empty:
            display_df = recommendations[['name', 'city', 'rating', 'price', 'cuisine', 'address']].copy()
            display_df.columns = ['Name', 'City', 'Rating', 'Cost (₹)', 'Cuisine', 'Address']
            st.dataframe(display_df, use_container_width=True, hide_index=True)
        else:
            st.info("Filter the data to see the comparison table.")

    with tab3:
        if not recommendations.empty:
            col_v1, col_v2 = st.columns(2)
            with col_v1:
                st.write("**Price vs Rating (Selection in Red)**")
                fig1, ax1 = plt.subplots()
                sns.scatterplot(data=df, x='price', y='rating_numeric', color='lightgrey', alpha=0.3, label='Full Market', ax=ax1)
                sns.scatterplot(data=recommendations, x='price', y='rating_numeric', color='red', s=80, label='Selected', ax=ax1)
                ax1.set_xlabel("Cost (₹)")
                ax1.set_ylabel("Rating")
                st.pyplot(fig1)

            with col_v2:
                st.write("**Rating Density Distribution**")
                fig2, ax2 = plt.subplots()
                sns.kdeplot(df['rating_numeric'].dropna(), fill=True, label="Market", color="grey", ax=ax2)
                if len(recommendations) > 1:
                    sns.kdeplot(recommendations['rating_numeric'].dropna(), fill=True, label="Selected", color="red", ax=ax2)
                plt.legend()
                st.pyplot(fig2)
        else:
            st.write("Not enough data for visual analysis.")

    # Footer
    st.sidebar.divider()
    st.sidebar.caption(f"Search results: {len(recommendations)} / {len(df)}")
```
